# 一、Number数字类型

#### 1.整数（int）
int表示正整数、负整数或零，不带小数点

```python
a = 1
print(type(a)) # type()查看变量a的类型
print(a)
```

```python
a = 1
b = 2
c = a+b # =是赋值运算符，==是比较运算符
d = a-b
e = a*b 
f = a/b # float类型，除法运算结果为浮点数
print(c, d, e, f)
print(type(f))
```

```python
# 整型的整除
c = 10//3 # //是整除运算符，结果为整数. 原本结果应为3.333..., 但整除运算符会舍弃小数部分，结果为3
print(c)
print(type(c)) 
```

```python
a = 10%3 # %是取余运算符，结果为1
print(a)
```

```python
a = 2**3 # **是幂运算符，结果为8
print(a)
```

#### 2.布尔型（bool）
值为True： 非0数、非空字符串、True  
值为False：0、空字符串、None、False  
bool有True和False，bool是int的子类，可与数字相加，True的值为1,False为0

```python
print(True)
print(True+1)
print(type(True))
print(False)
print(False+1)
```

```python
a = True + 1
b = False + 1
c = True - 1
print(a, b, c) # True在Python中等于1，False等于0，所以a的值为2，b的值为1，c的值为0
```

```python
bool("")
```

#### 3.浮点型（float）
浮点型由整数部分和小数部分组成。浮点型数据是近似存储的，可能存在精度误差。  
浮点型也可用科学计数法来表示，如1.8e3 = 1.8 * 10^3 = 1800

```python
a = 3.1415
b = int(a) # 将浮点型转换为整型，只保留整数部分
print(b, type(b))
```

```python
a = 3.1415926535
print(type(a))
print("a的值是%f" %(a))     # %f 默认保留两位小数
print("a的值是：%.3f" %(a)) # %.3f 保留3位小数
print(f"a的值是：{a}")      # 原样输出
print(f"a的值是：{a:.2f}")  # .2f 保留2位小数
```

```python
round(3.14159, 2) # round()函数用于对浮点数进行四舍五入，第二个参数表示保留的小数位数
```

```python
# 科学计数法
a = 1e3 # 1 * 10^3
print(a, type(a))
```

```python
b = 1e-3 # 1 * 10^(-3)
print(b)
```

#### 4.复数型(complex)
复数由实数部分和虚数部分组成，可用 a+bj 或 complex(a, b) 来表示。  
复数的实部a和虚部b都是浮点型

```python
# 复数类型
#c = 1+2j
c = complex(1, 2)
print(c, type(c))
```

# 二、String字符串类型
创建字符串可以使用单引号‘’、双引号“”、三单引号‘’‘  ’‘’、三双引号“”“  ”“”。其中，三引号可以多行定义字符串。   
字符串由 数字、字母、下划线组成。  
字符串索引：从左往右0、1、2、3、4 ； 从右往左：-5、-4、-3、-2、-1

```python
str1 = "hello world" # 双引号
print(str1)
print(type(str1))
```

```python
str2 = """hello world""" # 三双引号
print(str2)
print(type(str2))
```

```python
# 三双引号可以作注释使用
"""
112
234r444
wef
"""
print(12)
```

```python
# 字符串相加
a = "abc"
b = "def"
print(a+b)
```

```python
# 字符索引查询
str = "abcdefg"
print(str[1])
print(str[-1])
```

```python
# 字符串切片，基本语法：sequence(start:end:step)
str = "abcdefghijklnm"
print(str[::])    # 全输出
print(str[:])     # 全输出
print(str[1::])   # 从索引1开始，默认步长为1，左闭右开
print(str[2::2])  # 从索引1开始，步长为2，左闭右开
print(str[1:9:1]) # 从索引1开始8结束，步长为1
print(str[1:3])   # 从索引1开始2结束，默认步长为1
print(str[-10:-1]) # 从索引-10开始-2结束，默认步长为1
print(str[-10:])   # 从索引-10开始-2结束，默认步长为1，左闭右开
print(str[:-2])    # 从索引-2开始往左，默认步长为1，左开右闭
```

```python
# 字符串地址id
str = "abc"
print(str)
print(id(str))
str = "aaa"
print(str)
print(id(str)) # str重新赋值，会新存放一个地址
```

# 三、List列表
List（列表）是python中使用最频繁的数据类型  
列表是写在方括号[]之间、用逗号分隔开的元素列表，list中的元素类型可以不同，  
支持字符串、数字、甚至包含嵌套列表。列表中的元素可以改变。

```python
# 列表
lst1 = [1, 2, 3, 4, "hello", [1, 2, 3]]
print(lst1)
print(type(lst1))
```

```python
# 列表嵌套
a = [[1,2], [3,4]]
b = [[2,2],
     [2,2]]    
print(a) 
print(b)
```

```python
# 列表查询
lst1 = [1, 2, 3, 4, "hello", [1, 2, 3]]
print(lst1[1])
print(lst1[4])
print(type(lst1[-1]))
```

```python
# 列表切片
lst1 = [1, 2, 3, 4, "hello", [1, 2, 3]]
print(lst1[:])
print(lst1[1:])
print(lst1[1::2])
print(lst1[-4:])
```

```python
# 列表相加，列表不能相减
a = [1,2,3]
b = [3,4]
c = a + b
print(c)
```

```python
# list()创建列表函数
# list(range(start,end,step))
a = list(range(0,10))
print(a)
```

```python
a.append(10) # 增加列表元素
print(a)
```

```python
a.remove(0) # 删减列表元素
print(a)
```

```python
a[0] = "a" # 替换列表元素
print(a)
```

```python
list = [1,2,3,4]
print(max(list)) # 最大值
print(min(list)) # 最小值
print(sum(list)) # 求和
print(len(list)) # 求长度
```

```python
# 列表 升序 排序
list = [3,2,4,1]
list.sort()
print(list)
```

```python
# 列表 降序 排序
list = [3,2,4,1]
list.sort(reverse=True)
print(list)
```

# 四、Tuple元组
元组与列表相似，区别在于元组的元素不能修改  

```python
# 元组的定义：使用小括号()定义，而列表使用方括号[]
my_tuple = (1,2,3)
print(my_tuple)
print(type(tuple))
```

```python
# 元组只有一个元素，需要在后面加逗号，
b = (1,)     
print(type(b))
# 否则默认为整型
a = (1)
print(type(a))
```

```python
# 元组的索引
tuple = (1,2,3,4)
print(tuple[0])
print(tuple[-1])
# 元组的截取（切片）
print(tuple[1:3])
print(tuple[-3:-1])
```

```python
# .count() 元组的元素个数统计
a = (1,2,3,3,1,1,1)
print(a)
print(a.count(1)) # a.count(1) 统计元素1的个数
```

```python
# .index() 元素最先出现的位置（索引）
a = (0,2,1,3,1,1,1)
print(a.index(1))
```

# 五、Set集合
集合（Set）是一个无序、不重复元素的序列，即集合中不能出现重复元素，是数学中集合的体现。  

```python
# 集合的定义

# 1、使用大括号{}创建
my_set1 = {1,2,3}
print(my_set1)

# 2、使用 set() 函数创建
my_set2 = set([1,2,3,4])
print(my_set2)
```

```python
# 空集合只能用 set() 创建，因为{}是用来创建空字典的

# 正确创建空集合
empty_set = set()
print(type(empty_set))

# 错误创建空集合，{}是空字典
empty_dict = {}
print(type(empty_dict))
```

```python
# 集合的特点：无序、不重复

# 集合元素无先后顺序，不支持索引和切片

set1 = {1,2,2,3,4,4}
print(set1) # 输出{1,2,3,4}，重复元素自动去除
set2 = {3,4,4,2,2,1}
print(set2)
```

```python
a = {1,2,2,3}
print(a) # 自动去重
a.add(4) # 增加元素
print(a)
```

```python
a = {1,2,3}
a.remove(2) # 删除元素
print(a)
```

```python
a = {1,2,3}
del a # 删除整个集合
print(a)
```

```python
# 集合的交集、并集
a = {1,2,3,4}
b = {3,4,5,6}
c = a&b # 交集
d = a|b # 并集
print(c)
print(d)
```

# 六、Dictionary字典
字典是可变数据类型，可存储任意类型的对象。字典是一种映射类型，其元素是键值对（key-value pairs，即一个键对应一个值），  
○ 键：必须是不可变类型（如数字、字符串、元组等），且是唯一的  
○ 值：可变，可以是任何类型数据

```python
# 字典用{}创建，每个元素由键值对组成，键和值用冒号：分隔，键值对之间用逗号，分隔
a = {"name":"小明","age":18,"height":185}
print(a)
```

```python
# 增加一个键值对
a["IQ"] = 250
print(a)
```

```python
# 删除一个键值对
del a["IQ"]
print(a)
```

```python
# 该一个键对应的值
a["age"] = 23
print(a)
```

```python
# 查询键值
print(a["name"])
```

```python
# 键唯一，重复的键选最后一个
a = {"name":"小明","age":18,"height":185,"name":"小李"}
print(a)
```

```python
# 获取字典所有的键或值
my_dict = {"name":"小明","age":18,"height":185}
keys = my_dict.keys()    # 获取所有键
values = my_dict.values() # 获取所有值
print(keys)
print(values)
```
