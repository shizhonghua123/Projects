# 一、print输出函数
print(*values, sep=" ", end="\n", file=None, flush=False)

```python
# 双引号“”括起来的叫 字符串
# 双引号“”和单引号‘’功能相同
print("Hello World!")
print('Hello World!')
print("123") 
```

```python
print(123)  # 123是整数
print(12.3) # 12.3是浮点数
```

```python
print(1,2,3,4) # 无sep，默认分隔符sep为空格
print(1,2,3,4,sep="->") # 自定义分隔符sep为"->"
print(1,2,3,4,sep="$")
```

```python
# 打开一个文件，文件名为11.txt，模式为追加模式“a"（表示以写入方式打开文件并把写入位置放在文件末尾）
# 若文件不存在，则自动创建新文件
f = open("11.txt","a") 
print("Hello World!", file=f) # 将字符串"Hello World!"写入文件f中
f.close() # 关闭文件
```

```python
# 打开一个文件，文件名为22.txt，模式为写入模式“w"（表示以写入方式打开文件并把写入位置放在文件开头）
# 'w' 会清空原文件再写入
f = open("22.txt","w") 
print("Hello World!", file=f) # 将字符串"Hello World!"写入文件f中
f.close() # 关闭文件
```

```python
# 使用with语句打开文件"11.txt"，模式为追加模式“a",
with open("11.txt","a") as f:
    print("Hello World!", file=f)
# with语句会自动关闭文件
```

```python
with open("22.txt", "w") as f:
    print("1256", file=f)
```

# 二、格式化输出
含义：定义一个模板，按照模板输出内容
常用格式化字符：(1) %c : 字符（character）
             (2) %s : 字符串（string）
             (3) %d : 十进制整数（decimal）
             (4) %f : 浮点数（float）

```python
# 使用 % 格式化字符串，%s表示字符串，%d表示整数，需要指定类型
print("我的名字是%s, 年龄是%d岁" %("小明", 18))
```

```python
# 使用 format() 方法格式化字符串，不用指定类型，自动识别
print("我的名字是{}, 年龄为{}岁".format("小明", 18))
```

```python
# 定义变量name和age，并使用f-string格式化字符串输出
name, age = "小李", 20
# name = "小明"
# age = 18
print(f"我的名字是{name}, 年龄为{age}岁")
```

# 四、input()输入函数

```python
a = input("请输入一些内容：")
print("你输入的内容是：", a)
print(type(a)) # type()查看字符类型，input函数返回的内容类型为字符串
```
