# 三、循环结构

#### 1、While循环

```python
a = 0
while a<10:
    print(a)
    a+=1 # a = a + 1
```

```python
# 1到100的和
i = 0
sum = 0
while i<=100:
    sum+=i # sum = sum + i
    i+=1   # i = i + 1
print(sum)
```

```python
# 求1-100的偶数和
i = 0
sum = 0
while i<=100:
    sum+=i
    i+=2
print(sum)
```

```python
# 求1-100的偶数和
i = 0
sum = 0
while i<=100:
    if i%2==0:
        sum+=i
    i+=1
print(sum)
```

```python
# 求1-100的奇数和
i = 1
sum = 0
while i<=100:
    sum+=i
    i+=2
print(sum)
```

```python
# 打印九九乘法表
a = 1
while a<=9:
    b = 1
    while b<=a:
        print(f"{b}*{a}={a*b}", end="\t") # end="\t"，表示一个“tab”的间隔
        b+=1
    print() # 等价于print(end="\n") 换行
    a+=1
```

```python
i = 1
while i<=5:
    j = 1
    while j<=i:
        print("*", end=" ")
        j+=1
    print()
    i+=1
```

```python
while True: # 进入无限循环，直到break结束循环
    score = int(input("请输入成绩："))
    if score<0 or score>100:
        print("成绩输入错误，请重新输入")
        continue # 若输入无效，跳过当前循环，重新输入
    if score>=0 and score<60:
        print("不及格")
    elif score>=60 and score<80:
        print("普通")
    elif score>=80 and score<90:
        print("良好") 
    else:
        print("优秀")
    break # 结束循环
```

#### 2、For循环

```python
for i in range(10):
    print(i)
```

```python
# For循环打印九九乘法表
for i in range(1,10):
    j = 1
    for j in range(1,i+1):
        print(f"{j}*{i}={i*j}", end=("\t"))
    print()
```

#### 2、自定义函数
def 函数名(参数)：  
&emsp;&emsp;函数体  
&emsp;&emsp;return [表达式]

```python
# 定义加法函数
def add(a,b):
    sum = a+b
    return sum

print(add(1,2))
```
