# 面向对象编程
#### 1、面向过程编程
以过程为中心，把事情拆分为几个步骤，然后按顺序执行。
#### 2、面向对象编程
把事物抽象为对象，然后给每一个对象赋一些属性和方法，让每个对象都去执行自己的方法

#### 举个洗衣服的例子
-  面向过程：放衣服、放洗衣粉、放水、清洗、晾干    
-  面向对象：  
1、创建出两个对象：“人”和“洗衣机”   
2、在“人”和“洗衣机”中加入属性和方法   
3、然后只需执行这两个对象即可  

#### 类和对象
-  类的定义  
类是具有相同属性和功能的一类事物  
-  对象的定义  
对象是类的具体体现，是面向对象编程的核心。例如：   
动物（类） → 猫、狗、熊（对象）   
人类（类） → 张三、李四、王五（对象）  

```python
a = 10
b = 9.9
c = "abc"
print(type(a))
print(type(b))
print(type(c))
# type是类(class), a、b、c是对象
```
